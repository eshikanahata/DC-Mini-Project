```python
import tensorflow as tf
import tensorflow_hub as hub
import pandas as pd
import numpy as np
```

```python
words = ['apple', 'guava', 'watermelon', 'banana', 'pear', 'Google', 'Microsoft', 'Oracle', 'Meta', 'Nvidia', 'Pink Floyd', 'Led Zeppelin', 'Dire Straits', 'The Doors', 'The Beatles', 'chicken', 'ball', 'guitar', 'piano', 'water']
```

```python
hub = hub.load("https://tfhub.dev/google/universal-sentence-encoder/4")
```

```python
embeddings = hub(words)
embeddings = embeddings.numpy()
```

```python
pd.DataFrame(embeddings).to_csv('vectors.tsv', sep='\t', index=False, header = False)
pd.DataFrame(words).to_csv('metadata.tsv', sep='\t', index =False, header = False)
```
